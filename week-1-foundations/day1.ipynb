Filtering

```python
# Convert columns to numeric (only if needed)
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce")

# Products with unit_price > 1000
df[df["unit_price"] > 1000]
```

```python
# Products with quantity >= 5
df[df["quantity"] >= 5]
```

```python
# Products satisfying both conditions
df[(df["unit_price"] > 1000) & (df["quantity"] >= 5)]
```

Sorting

```python
edf_copy = df.sort_values(by="unit_price")    # Ascending
df_desc = df.sort_values(by="unit_price", ascending=False)   # Descending
display(df)
display(df_copy)
display(df_desc)
```
