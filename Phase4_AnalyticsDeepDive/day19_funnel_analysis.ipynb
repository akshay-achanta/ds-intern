# Day 19: Funnel Analysis
User funnel chart with stage-by-stage drop-off rates. Since we lack web event data, we simulate a checkout funnel.

```python
import pandas as pd
import plotly.express as px

# Simulated funnel data
data = dict(
    stage=['Website Visit', 'Add to Cart', 'Checkout', 'Purchase'],
    users=[10000, 4500, 2000, 850]
)
funnel_df = pd.DataFrame(data)
```

## Funnel Visualization

```python
fig = px.funnel(funnel_df, x='users', y='stage', title='E-commerce Conversion Funnel')
fig.show()
```

## Observations
1. **High Drop-off at Add to Cart:** Only 45% of visitors add an item to their cart. We need to improve product pages.
2. **Checkout Abandonment:** From Checkout to Purchase, there is a significant drop. This indicates friction in the payment process.
